### 3.1 实验2 操作任务1：score_level

编写 `score_level(score)` 函数：

- 90~100 返回“优秀”
- 80~不足90 返回“良好”
- 60~不足80 返回“及格”
- 0~不足60 返回“不及格”
- 超出 0~100 范围抛出 `ValueError`

```python
def score_level(score: float) -> str:
    """
    根据分数返回等级。

    参数:
        score: 分数，0~100

    返回:
        "优秀" / "良好" / "及格" / "不及格"

    异常:
        ValueError: 分数超出 0~100
    """
    if score < 0 or score > 100:
        raise ValueError("分数必须在 0~100 范围内")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

# 测试正常分数
print(score_level(95))
print(score_level(85))
print(score_level(70))
print(score_level(50))

# 测试越界分数，捕获异常
try:
    score_level(105)
except ValueError as e:
    print("异常：", e)
```

**解释：**

- 先判断分数是否在 0~100，越界则抛出 `ValueError`。
- 使用 `if-elif-else` 按区间返回对应等级。
- 测试 95、85、70、50 分别返回优秀、良好、及格、不及格。
- 测试 105 抛出异常并被捕获。

### 3.2 实验2 操作任务2：for 与 while 计算 1 到 100 的和

分别用 `for` 和 `while` 计算 1 到 100 的整数和，并比较结果。

```python
# for 循环求和
sum_for = 0
for i in range(1, 101):
    sum_for += i

# while 循环求和
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1

print("for 结果:", sum_for)
print("while 结果:", sum_while)
```

**比较结果：**

- **结果相同**：两种循环计算 1 到 100 的整数和，输出均为 `5050`。
- **验证**：等差数列求和公式 \( S = \frac{100 \times 101}{2} = 5050 \)，与两种实现结果一致。
- **for 循环**：适合循环次数已知或可迭代的场景，代码更简洁，循环变量自动生成，不易出错。
- **while 循环**：适合循环次数不确定、依赖条件判断的场景，需要手动初始化并更新循环变量，容易漏写自增导致死循环。
- **结论**：对于已知范围的求和，`for` 循环更直观、更安全；`while` 循环也能实现，但代码稍繁琐。

### 3.3 实验2 操作任务3：递归版和迭代版阶乘

约定 `0! = 1`，输入为非负整数。比较 `n = 0、1、5、10` 时的结果。

```python
# 递归版阶乘
def factorial_recursive(n: int) -> int:
    """
    用递归方式计算 n 的阶乘。

    参数:
        n: 非负整数

    返回:
        n!，约定 0! = 1
    """
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

# 迭代版阶乘
def factorial_iterative(n: int) -> int:
    """
    用迭代方式计算 n 的阶乘。

    参数:
        n: 非负整数

    返回:
        n!，约定 0! = 1
    """
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# 比较 n = 0, 1, 5, 10 的结果
for n in [0, 1, 5, 10]:
    print(n, factorial_recursive(n), factorial_iterative(n))
```

**比较结果：**

- `n = 0`：递归版和迭代版都返回 `1`，符合约定 `0! = 1`。
- `n = 1`：两者都返回 `1`。
- `n = 5`：两者都返回 `120`。
- `n = 10`：两者都返回 `3628800`。
- **递归版**：代码更简洁，直接表达阶乘的数学定义；但递归深度过大时可能触发 Python 递归深度限制。
- **迭代版**：使用循环累乘，更稳定，适合较大的 `n`。
- **结论**：两种实现结果完全一致，递归适合理解定义，迭代更适合实际工程。

### 3.4 实验2 操作任务4：summary(values)

接收非空数值列表，以元组返回最小值、最大值和平均值，并通过元组解包接收结果。

```python
from typing import Sequence

def summary(values: Sequence[float]) -> tuple:
    """
    返回数值序列的最小值、最大值和平均值。

    参数:
        values: 非空数值序列

    返回:
        (最小值, 最大值, 平均值)

    异常:
        ValueError: values 为空
    """
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)

# 调用函数并用元组解包接收结果
min_v, max_v, avg_v = summary([85, 58, 92, 76, 60, 89])
print("最小值:", min_v)
print("最大值:", max_v)
print("平均值:", avg_v)
```

**解释：**

- 函数返回一个元组：`(最小值, 最大值, 平均值)`。
- 通过元组解包 `min_v, max_v, avg_v = summary(...)` 一次性接收三个结果。
- 输入列表为 `[85, 58, 92, 76, 60, 89]`：
  - 最小值：`58`
  - 最大值：`92`
  - 平均值：`460 / 6 ≈ 76.666...`
- 如果传入空列表，会抛出 `ValueError`。

### 3.5 实验2 操作任务5：sum_all(*args)

返回所有参数之和，分别测试无参数、单个参数和多个参数的情况。

```python
def sum_all(*args: float) -> float:
    """
    接收任意数量参数，返回它们的和。

    参数:
        *args: 任意数量的数值参数

    返回:
        所有参数之和；无参数时返回 0
    """
    return sum(args)

# 测试无参数、单参数、多参数
print(sum_all())
print(sum_all(5))
print(sum_all(1, 2, 3, 4, 5))
```

**解释：**

- `*args` 接收任意数量的位置参数，并将它们打包成一个元组。
- `sum(args)` 对元组中的所有元素求和。
- 测试结果：
  - `sum_all()`：无参数，元组为空，返回 `0`。
  - `sum_all(5)`：单个参数，返回 `5`。
  - `sum_all(1, 2, 3, 4, 5)`：多个参数，返回 `15`。
- **结论**：`*args` 让函数可以处理可变数量的参数。

### 5.3 实验4 操作任务3：编写 parse_student 函数

将成绩转换为整数，检查字段数量、姓名是否为空及成绩是否在 0~100 范围内。无效记录抛出 `ValueError`。

```python
def parse_student(line: str) -> dict:
    """
    解析一行学生记录。

    参数:
        line: 形如 "Alice,85" 的字符串

    返回:
        {"name": str, "score": int}

    异常:
        ValueError: 字段数量不对、姓名为空、成绩非整数或超出 0~100
    """
    # 去掉首尾空白，按逗号拆分
    parts = line.strip().split(",")

    # 检查字段数量是否为 2
    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}")

    name = parts[0].strip()
    score_text = parts[1].strip()

    # 检查姓名是否为空
    if not name:
        raise ValueError("姓名为空")

    # 尝试把成绩转成整数
    try:
        score = int(score_text)
    except ValueError:
        raise ValueError(f"成绩不是整数: {score_text!r}")

    # 检查成绩范围
    if score < 0 or score > 100:
        raise ValueError(f"成绩超出 0~100: {score}")

    return {"name": name, "score": score}

# 测试有效和无效输入
print(parse_student("Alice,85"))

try:
    parse_student("Grace,abc")
except ValueError as e:
    print("无效输入测试:", e)
```

**解释：**

- `line.strip().split(",")` 去掉首尾空白并按逗号拆分字段。
- 字段数量不是 2 时抛 `ValueError`。
- 姓名空、成绩不是整数、成绩超出 0~100 都会抛 `ValueError`。
- 函数返回字典，方便后续保存和处理。

### 5.4 实验4 操作任务4：按行处理文件，捕获无效记录

按行处理文件，捕获无效记录的异常，记录行号与错误原因，并继续处理后续记录。有效记录保存为字典组成的列表。

```python
valid_records = []   # 保存有效记录
invalid_records = [] # 保存无效记录及错误信息

with open(data_path, "r", encoding="utf-8") as f:
    for lineno, line in enumerate(f, start=1):
        # 跳过头行
        if lineno == 1:
            continue
        # 跳过空行
        if not line.strip():
            continue

        try:
            record = parse_student(line)
            valid_records.append(record)
        except ValueError as e:
            invalid_records.append({
                "line_no": lineno,
                "line": line.strip(),
                "error": str(e)
            })
            print(f"第 {lineno} 行无效: {line.strip()} -> {e}")

print("\n有效记录:", valid_records)
print("无效记录:", invalid_records)
```

**解释：**

- `enumerate(f, start=1)` 同时获取行号和行内容。
- 第 1 行是表头，跳过。
- 每行调用 `parse_student()`，成功则加入 `valid_records`，失败则捕获异常并加入 `invalid_records`。
- 程序继续处理后续行，不会因某一行出错而中断。
- 有效记录共 6 条，无效记录共 3 条（Grace,abc；Henry,105；Incomplete）。

```python
from collections import Counter

error_types = Counter()
for r in invalid_records:
    if "字段数量" in r["error"]:
        error_types["字段数量错误"] += 1
    elif "不是整数" in r["error"]:
        error_types["成绩非整数"] += 1
    elif "超出" in r["error"]:
        error_types["成绩越界"] += 1
    elif "姓名为空" in r["error"]:
        error_types["姓名为空"] += 1

print("无效记录分类统计:")
for k, v in error_types.items():
    print(f"  {k}: {v} 条")
```

**说明：**

- 3 条无效记录中：
  - 1 条字段数量错误（Incomplete）
  - 1 条成绩非整数（Grace,abc）
  - 1 条成绩越界（Henry,105）
- 分类统计有助于快速定位数据质量问题。

### 5.5 实验4 操作任务5：写入 report.txt，演示 w 与 a 模式

将有效记录数、无效记录数、平均分和及格率写入 `exp01_data/report.txt`。分别演示 `w` 与 `a` 模式，说明二者对已有内容的影响。

```python
# 计算统计指标
valid_count = len(valid_records)
invalid_count = len(invalid_records)

if valid_count > 0:
    avg_score = sum(r["score"] for r in valid_records) / valid_count
    pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / valid_count
else:
    avg_score = 0.0
    pass_rate = 0.0

# 生成报告文本
report_text = (
    f"有效记录数: {valid_count}\n"
    f"无效记录数: {invalid_count}\n"
    f"平均分: {avg_score:.2f}\n"
    f"及格率: {pass_rate:.2%}\n"
)

report_path = os.path.join(data_dir, "report.txt")

# 使用 w 模式写入：会覆盖原有内容
with open(report_path, "w", encoding="utf-8") as f:
    f.write(report_text)

print("w 模式写入后:")
print(report_text)
```

**比较结果：**

- `w` 模式：打开文件时清空原有内容，再写入新内容。
- `a` 模式：保留原有内容，在文件末尾追加新内容。

## 6. 实验5：类、继承与对象复制

### 6.1 实验5 操作任务1：定义 Student 类

定义 `Student` 类，在 `__init__()` 中初始化 `name` 和 `score` 属性，并实现 `is_passed()` 和 `__str__()` 方法。

```python
class Student:
    """学生类，包含姓名和成绩"""

    def __init__(self, name: str, score: int) -> None:
        self.name = name
        self.score = score

    def is_passed(self) -> bool:
        """成绩大于等于 60 视为及格"""
        return self.score >= 60

    def __str__(self) -> str:
        """面向用户的字符串表示，print(obj) 时调用"""
        return f"{self.name}({self.score})"

    def __repr__(self) -> str:
        """面向开发者的字符串表示，交互环境直接输入 obj 时调用"""
        return f"Student(name={self.name!r}, score={self.score!r})"

# 测试
s = Student("Alice", 85)
print("str:", str(s))
print("repr:", repr(s))
print("直接 print:", s)
print("列表中的显示:", [s])
```

**解释：**

- `__init__()` 是构造方法，创建对象时自动调用，用来初始化属性。
- `is_passed()` 用 `score >= 60` 判断是否及格。
- `__str__()` 决定 `print(obj)` 时的显示内容。

**`__str__` 与 `__repr__` 区别：**

- `__str__`：面向用户，`print(obj)` 时调用，输出简洁。
- `__repr__`：面向开发者，直接在交互环境输入 `obj` 时调用，输出应尽量能重建对象。
- 两者都实现时，`print()` 用 `__str__`，列表打印用 `__repr__`。

为实验4中的每条有效记录创建一个 `Student` 对象，输出对象信息和及格状态。

**解释：**

- `print(s)` 会调用 `__str__()` 方法。
- `s.is_passed()` 返回布尔值，配合条件表达式输出“及格”或“不及格”。

### 6.3 实验5 操作任务3：GraduateStudent 继承

定义继承自 `Student` 的 `GraduateStudent` 类，增加 `research_topic` 属性，通过 `super().__init__()` 初始化父类属性，并重写 `__str__()`。

```python
class GraduateStudent(Student):
    """研究生类，继承自 Student"""

    def __init__(self, name: str, score: int, research_topic: str) -> None:
        # 调用父类构造方法初始化 name 和 score
        super().__init__(name, score)
        # 新增研究方向属性
        self.research_topic = research_topic

    def __str__(self) -> str:
        """重写 __str__，增加研究方向信息"""
        return (
            f"GraduateStudent(name={self.name}, score={self.score}, "
            f"research_topic={self.research_topic})"
        )

# 创建研究生对象并测试
grad = GraduateStudent("Alice", 85, "人工智能")
print(grad)
print("是否及格:", grad.is_passed())
```

**解释：**

- `class GraduateStudent(Student)` 表示继承自 `Student`。
- `super().__init__(name, score)` 调用父类构造方法，避免重复写初始化逻辑。
- 新增 `research_topic` 属性表示研究方向。
- 重写 `__str__()` 让输出包含研究方向。
- `grad.is_passed()` 直接复用父类方法。

### 6.4 实验5 操作任务4：直接赋值、浅拷贝、深拷贝

构造包含嵌套列表的对象，例如 `{"name": "Alice", "scores": [85, 90]}`。分别进行直接赋值、`copy.copy()` 和 `copy.deepcopy()`，修改内部列表，比较结果。每次比较都从新建的原始对象开始。

```python
import copy

# ===== 直接赋值 =====
orig1 = {"name": "Alice", "scores": [85, 90]}
assigned = orig1                      # 直接赋值，指向同一个对象
assigned["scores"].append(100)        # 修改内部列表

print("直接赋值:")
print("orig1:   ", orig1)
print("assigned:", assigned)
print()

# ===== 浅拷贝 =====
orig2 = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(orig2)            # 浅拷贝，外层新对象，内层仍共享
shallow["scores"].append(100)         # 修改内部列表

print("浅拷贝:")
print("orig2:  ", orig2)
print("shallow:", shallow)
print()

# ===== 深拷贝 =====
orig3 = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(orig3)           # 深拷贝，内外层都独立
deep["scores"].append(100)            # 修改内部列表

print("深拷贝:")
print("orig3:", orig3)
print("deep: ", deep)
```

**比较结果：**

- **直接赋值**：`assigned = orig1` 让两个变量指向同一个对象。修改 `assigned["scores"]` 后，`orig1["scores"]` 也跟着变，说明是同一对象。
- **浅拷贝**：`copy.copy()` 创建新的外层字典，但内层列表仍然共享。修改 `shallow["scores"]` 后，`orig2["scores"]` 也跟着变。
- **深拷贝**：`copy.deepcopy()` 递归复制所有嵌套对象。修改 `deep["scores"]` 后，`orig3["scores"]` 不受影响。

**结论：**

| 方式 | 外层对象 | 内层对象 | 修改内层是否影响原对象 |
|---|---|---|---|
| 直接赋值 | 共享 | 共享 | 是 |
| 浅拷贝 | 独立 | 共享 | 是 |
| 深拷贝 | 独立 | 独立 | 否 |

- 需要完全独立副本时，用 `copy.deepcopy()`。
- 只需要外层独立、内层可共享时，用 `copy.copy()`。
- 直接赋值不创建新对象，只是多了一个引用。
