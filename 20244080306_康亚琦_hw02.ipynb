## 2. 字符串的乐趣

### 2.1 判断回文 (is_palindrome)

**任务要求：** 忽略空格和大小写，判断字符串是否为回文。

```python
def is_palindrome(s: str) -> bool:
    """
    判断字符串是否为回文（忽略空格和大小写）。
    
    参数:
        s: 输入字符串
    返回:
        True 如果是回文，否则 False
    """
    # 去空格并转为小写
    clean_s = s.replace(" ", "").lower()
    # 比较正序和倒序是否一致
    return clean_s == clean_s[::-1]

# 测试
print("is_palindrome('taco cat'):", is_palindrome("taco cat"))
print("is_palindrome('T A C O cat'):", is_palindrome("T A C O cat"))
print("is_palindrome('hello'):", is_palindrome("hello"))
```

**关键结果解释：**
- 运行结果输出了 `True, True, False`。通过 `replace(" ", "")` 去除空格，`lower()` 统一小写后，正序字符串与倒序切片 `[::-1]` 完全相等。说明函数成功过滤了格式干扰，实现了回文判断。

### 2.2 判断字母顺序 (is_abecedarian)

**任务要求：** 忽略空格，判断单词的字母是否按从左到右的字母顺序排列。

```python
def is_abecedarian(s: str) -> bool:
    """
    判断字符串的字母是否按字母顺序排列（忽略空格）。
    
    参数:
        s: 输入字符串
    返回:
        True 如果是字母顺序，否则 False
    """
    clean_s = s.replace(" ", "").lower()
    # 遍历比较相邻字符
    for i in range(len(clean_s) - 1):
        if clean_s[i] > clean_s[i + 1]:
            return False
    return True

# 测试
print("is_abecedarian('adder'):", is_abecedarian("adder"))
print("is_abecedarian('abcd efgh xyz'):", is_abecedarian("abcd efgh xyz"))
print("is_abecedarian('dog'):", is_abecedarian("dog"))
```

**关键结果解释：**
- 运行结果输出了 `True, True, False`。`abcd efgh xyz` 去空格后相邻字母对比均未出现降序情况，而 `dog` 出现了 `o > g`，返回 `False`。函数通过单次遍历判断相邻字符大小，符合逻辑。

### 2.3 移除元音 (remove_vowels)

**任务要求：** 去除字符串中的所有元音字母（a, e, i, o, u）。

```python
def remove_vowels(s: str) -> str:
    """
    移除字符串中的所有元音 (a, e, i, o, u)。
    
    参数:
        s: 输入字符串
    返回:
        去除元音后的新字符串
    """
    vowels = "aeiouAEIOU"
    return "".join([char for char in s if char not in vowels])

# 测试
print("remove_vowels('cat'):", remove_vowels("cat"))
print("remove_vowels('audio'):", remove_vowels("audio"))
print("remove_vowels('AEIOU'):", remove_vowels("AEIOU"))
```

**关键结果解释：**
- 运行结果成功将 `'cat'` 转为 `'ct'`，`'audio'` 转为 `'d'`，全元音字符串 `'AEIOU'` 输出为空字符串。说明列表推导式成功过滤了指定的元音集合。

## 3. 列表的乐趣

### 3.1 反转列表 (list_reverse)

**任务要求：** 返回反转后的列表。如果输入不是列表，引发适当的错误。

```python
def list_reverse(lst: list) -> list:
    """
    返回反转后的列表。
    
    参数:
        lst: 输入列表
    返回:
        反转后的新列表
    异常:
        TypeError: 如果输入不是列表
    """
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]

# 测试
print("list_reverse([1, 2, 3]):", list_reverse([1, 2, 3]))

try:
    list_reverse("abc")
except TypeError as e:
    print("捕获异常:", e)
```

**关键结果解释：**
- 传入 `[1, 2, 3]` 时成功返回 `[3, 2, 1]`；传入字符串 `"abc"` 时，`isinstance` 检查生效，抛出 `TypeError` 并被捕获。说明类型检查和反转逻辑正常。

### 3.2 判断是否升序 (is_sorted)

**任务要求：** 判断序列是否按升序排列，必须只遍历一次。

```python
def is_sorted(seq) -> bool:
    """
    判断序列是否按升序排列（仅遍历一次）。
    
    参数:
        seq: Python 序列
    返回:
        True 如果升序，否则 False
    """
    # 使用 zip 将原序列和切片后的序列错位配对，单次遍历比较
    return all(seq[i] <= seq[i + 1] for i in range(len(seq) - 1))

# 测试
print("is_sorted([1, 2, 2, 3]):", is_sorted([1, 2, 2, 3]))
print("is_sorted([1, 3, 2]):", is_sorted([1, 3, 2]))
```

**关键结果解释：**
- 运行结果输出 `True` 和 `False`。代码使用生成器表达式 `range(len(seq) - 1)` 配合 `all()`，保证了对序列的**单次线性遍历**，没有使用 `sorted()` 或嵌套循环，完全符合题目“高效解法”的要求。

## 4. 作为向量的元组

### 4.1 标量乘法 (vec_scalar_mult)

**任务要求：** 向量（元组）与标量相乘。不合法的输入类型引发 TypeError。

```python
def vec_scalar_mult(t: tuple, s) -> tuple:
    """
    向量与标量相乘。
    
    参数:
        t: 数值元组
        s: 标量数值
    返回:
        相乘后的新元组
    """
    if not isinstance(t, tuple) or not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("第一个参数必须是数值元组")
    if not isinstance(s, (int, float)):
        raise TypeError("第二个参数必须是整数或浮点数")
    
    return tuple(x * s for x in t)

# 测试
print("vec_scalar_mult((1, 2, 3), 2):", vec_scalar_mult((1, 2, 3), 2))
```

**关键结果解释：**
- 运行结果为 `(2, 4, 6)`。通过 `isinstance` 检查了元组及标量的合法性，利用生成器表达式对每个元素乘以标量，返回了同长度的新元组。

### 4.2 内积 (vec_inner_product)

**任务要求：** 计算两个向量的内积。非元组引发 TypeError，长度不等引发 ValueError。

```python
def vec_inner_product(t1: tuple, t2: tuple) -> float:
    """
    计算两个向量的内积。
    
    参数:
        t1: 第一个数值元组
        t2: 第二个数值元组
    返回:
        内积浮点数
    """
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量的长度必须一致")
    
    return float(sum(x * y for x, y in zip(t1, t2)))

# 测试
print("vec_inner_product((1, 2), (3, 4)):", vec_inner_product((1, 2), (3, 4)))
```

**关键结果解释：**
- 运行结果为 `11.0`（1*3 + 2*4）。函数严格检查了输入的类型和长度，利用 `zip` 将对应元素配对相乘并求和，最终转换为浮点数输出。

### 4.3 检查矩阵有效性 (check_valid_mx)

**任务要求：** 矩阵为元组的元组，所有内层元组等长，且元素均为数值。有效返回 True，否则 False。

```python
def check_valid_mx(mx: tuple) -> bool:
    """
    检查输入是否为有效的矩阵。
    
    参数:
        mx: 待检查的元组
    返回:
        True 如果有效，否则 False
    """
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    
    # 检查每一行是否都是元组
    if not all(isinstance(row, tuple) for row in mx):
        return False
        
    # 检查列数是否一致
    row_len = len(mx[0])
    if not all(len(row) == row_len for row in mx):
        return False
        
    # 检查元素是否全为数值
    return all(isinstance(x, (int, float)) for row in mx for x in row)

# 测试
print("check_valid_mx(((1, 2), (3, 4))):", check_valid_mx(((1, 2), (3, 4))))
print("check_valid_mx(((1, 2), (3,))):", check_valid_mx(((1, 2), (3,))))
print("check_valid_mx(((1, 2), (3, 'a'))):", check_valid_mx(((1, 2), (3, 'a'))))
```

**关键结果解释：**
- 运行结果输出 `True, False, False`。第一项是标准二维矩阵；第二项内层长度不一致；第三项包含了非法字符 `'a'`。说明函数成功校验了矩阵的嵌套结构和数据类型。

### 4.4 矩阵乘向量 (mx_vec_mult)

**任务要求：** 矩阵与向量相乘，返回新元组。无效输入抛出适当异常。

```python
def mx_vec_mult(mx: tuple, vec: tuple) -> tuple:
    """
    矩阵乘以向量。
    
    参数:
        mx: 有效矩阵
        vec: 数值向量元组
    返回:
        结果向量元组
    """
    if not check_valid_mx(mx):
        raise ValueError("无效的矩阵输入")
    if not isinstance(vec, tuple) or not all(isinstance(x, (int, float)) for x in vec):
        raise TypeError("向量必须是数值元组")
    if len(mx[0]) != len(vec):
        raise ValueError("矩阵的列数必须等于向量的维度")
    
    # 矩阵的每一行与向量计算内积
    return tuple(sum(row[i] * vec[i] for i in range(len(vec))) for row in mx)

# 测试
print("mx_vec_mult(((1, 2), (3, 4)), (5, 6)):", mx_vec_mult(((1, 2), (3, 4)), (5, 6)))
```

**关键结果解释：**
- 运行结果为 `(17, 39)`（第一行：1*5+2*6=17，第二行：3*5+4*6=39）。函数先校验了矩阵，再校验向量长度，最后利用嵌套生成器完成矩阵乘法运算。

## 5. 更多向量的乐趣（类）

### 5.1 定义 Vector 类与属性说明

**任务要求：** 定义 `Vector` 类，包含维度与元素，进行类型与长度校验。
**问题回答：** 维度和元素应是**实例属性**，因为每个具体的向量对象都有独立的维度和数据，如果作为类属性，所有实例将共享同一份数据，会导致严重的逻辑错误。

```python
class Vector:
    """表示一个数学向量的类"""
    
    def __init__(self, dim: int, elements: list | tuple = None) -> None:
        if not isinstance(dim, int) or dim < 0:
            raise TypeError("维度必须是非负整数")
        
        if elements is None:
            elements = [0] * dim  # 默认行为：全 0 向量
            
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
            
        if len(elements) != dim:
            raise ValueError(f"维度 {dim} 与元素个数 {len(elements)} 不一致")
            
        if not all(isinstance(x, (int, float)) for x in elements):
            raise TypeError("元素必须全为数值")
            
        self.dim = dim
        self.elements = tuple(elements)
```

**关键结果解释：**
- 定义了 `Vector` 类并在 `__init__` 中完成了严格校验。如果只传维度，自动生成对应长度的全 0 向量；同时统一将元素转换为元组以保证不可变性。

### 5.2 比较运算符

**任务要求：** 实现 `==`、`<`、`<=`、`>` 等比较运算符。维度不同抛出异常。

```python
class Vector(Vector):
    def __eq__(self, other) -> bool:
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements == other.elements

    def __lt__(self, other) -> bool:
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements < other.elements
        
    def __le__(self, other) -> bool:
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements <= other.elements

    def __gt__(self, other) -> bool:
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements > other.elements
        
    def __ge__(self, other) -> bool:
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements >= other.elements
```

**关键结果解释：**
- 通过重写 Python 的魔法方法（`__lt__` 等），使得 `Vector` 对象可以直接使用比较运算符。如果在比较时发现维度不同，按题目要求主动抛出 `ValueError`；否则利用元组的字典序比较规则返回布尔值。

### 5.3 内积运算与乘法运算符

**任务要求：** 实现 `dot` 方法和乘法运算符 `*`（包含标量乘和向量乘）。

```python
class Vector(Vector):
    def dot(self, other: 'Vector') -> float:
        """计算两个向量的内积"""
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 对象")
        if self.dim != other.dim:
            raise ValueError("维度不一致")
        return sum(x * y for x, y in zip(self.elements, other.elements))

    def __mul__(self, other):
        # 向量 * 标量 或 向量 * 向量
        if isinstance(other, (int, float)):
            return Vector(self.dim, [x * other for x in self.elements])
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("两个向量维度不一致")
            return Vector(self.dim, [x * y for x, y in zip(self.elements, other.elements)])
        else:
            return NotImplemented

    def __rmul__(self, other):
        # 标量 * 向量 (如 2 * v)
        if isinstance(other, (int, float)):
            return self.__mul__(other)
        return NotImplemented

    def __repr__(self) -> str:
        return f"Vector({self.dim}, {self.elements})"

# 测试代码
v1 = Vector(2, (2, 3))
v2 = Vector(2, (4, 5))
print("v1 == v2:", v1 == v2)
print("v1 < Vector(2, (2, 5)):", v1 < Vector(2, (2, 5)))
print("v1.dot(v2):", v1.dot(v2))
print("2 * v1:", 2 * v1)
print("v1 * 2:", v1 * 2)
print("v1 * v2:", v1 * v2)
```

**关键结果解释：**
- `dot` 方法和 `__mul__` 都严格进行了维度和类型检查。
- 通过 `__mul__` 判断参数是标量还是向量，分别返回对应的新 `Vector` 对象。
- 额外实现了 `__rmul__`，使得 `2 * v`（标量在左）能正常运算，输出结果验证了对应元素相乘的逻辑正确。
