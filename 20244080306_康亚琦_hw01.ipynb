## 2. 热身：定义简单函数

1. 定义 `say_hi()`，不接受参数，打印 `"Hello, world!"`。
2. 定义 `goat_pad(s)`，接受一个字符串，打印前后都加上 `"goat"` 的字符串。
3. 定义 `print_n(s, n)`，接受字符串 `s` 和整数 `n`，打印 `s` 共 `n` 次，每次单独一行。

```python
# 1. 无参数函数，打印问候语
def say_hi() -> None:
    """打印问候语 Hello, world!"""
    print("Hello, world!")

# 2. 字符串前后加 "goat"
def goat_pad(s: str) -> None:
    """
    在字符串前后都加上 "goat"。

    参数:
        s: 输入字符串
    """
    print("goat" + s + "goat")

# 3. 打印字符串 n 次，每次一行
def print_n(s: str, n: int) -> None:
    """
    打印字符串 s 共 n 次，每次单独一行。

    参数:
        s: 要打印的字符串
        n: 打印次数
    """
    for _ in range(n):
        print(s)

# 测试
say_hi()
goat_pad("hello")
print_n("Python", 3)
```

**解释：**

- `say_hi()` 没有参数，直接打印。
- `goat_pad(s)` 用 `+` 拼接字符串。
- `print_n(s, n)` 用 `for _ in range(n)` 循环 n 次，每次打印一行。

## 3. 欧几里得算法

1. 实现 `gcd(m, n)`，返回两个整数的最大公约数。
2. 计算：
   - `gcd(2018, 2019)`
   - `gcd(1200, 300)`
   - `gcd(5040, 60)`
3. 如果一个或两个参数是负数，函数会做什么？这种行为有意义吗？

```python
def gcd(m: int, n: int) -> int:
    """
    使用欧几里得算法求两个整数的最大公约数。

    参数:
        m: 第一个整数
        n: 第二个整数

    返回:
        非负最大公约数
    """
    while n != 0:
        m, n = n, m % n
    return abs(m)  # 保证返回非负最大公约数

# 计算三组数的 GCD
print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300) =", gcd(1200, 300))
print("gcd(5040, 60) =", gcd(5040, 60))

# 测试负数
print("gcd(-2018, 2019) =", gcd(-2018, 2019))
print("gcd(2018, -2019) =", gcd(2018, -2019))
```

**结果：**

- `gcd(2018, 2019) = 1`
- `gcd(1200, 300) = 300`
- `gcd(5040, 60) = 60`

**负数行为：**

- 函数使用了 `abs(m)`，所以即使传入负数，也会返回非负的最大公约数。
- 数学上，最大公约数通常定义为非负整数，因此返回正数是有意义的。
- 如果不加 `abs`，Python 的 `%` 可能产生负数，导致结果可能为负，这不符合通常的数学定义。

## 4. 近似欧拉数 e

1. 定义 `euler_limit(n)`，用公式 `(1 + 1/n)^n` 近似 e。
2. 定义 `euler_infinite_sum(n)`，用公式（1）前 n 项之和近似 e。
   - `euler_infinite_sum(4)` 应返回前四项和：`1 + 1 + 1/2 + 1/6 ≈ 2.667`。
   - `euler_infinite_sum(1)` 返回 `1`。
   - `euler_infinite_sum(0)` 返回 `0`。
3. 定义 `euler_approx(epsilon)`，用 while 循环累加，直到与真实 e 的误差小于 `epsilon`。
4. 定义 `print_euler_sum_table(n)`，打印 `euler_infinite_sum(k)`，k 从 1 到 n，每行一个。

```python
import math

def euler_limit(n: int) -> float:
    """
    用极限公式 (1 + 1/n)^n 近似 e。

    参数:
        n: 正整数

    返回:
        e 的近似值
    """
    return (1 + 1 / n) ** n

def euler_infinite_sum(n: int) -> float:
    """
    用前 n 项和近似 e：sum(1/k!) for k in [0, n)。

    参数:
        n: 非负整数，表示项数

    返回:
        e 的近似值
    """
    total = 0.0
    for k in range(n):          # k = 0, 1, ..., n-1
        total += 1 / math.factorial(k)
    return total

def euler_approx(epsilon: float, max_iter: int = 1000) -> float:
    """
    用 while 循环累加前 n 项和，直到与真实 e 的误差小于 epsilon。

    参数:
        epsilon: 允许误差
        max_iter: 最大迭代次数，防止死循环

    返回:
        e 的近似值

    异常:
        RuntimeError: 达到最大迭代次数仍未满足误差
    """
    true_e = math.exp(1)
    total = 0.0
    k = 0
    while k < max_iter:
        total += 1 / math.factorial(k)
        k += 1
        if abs(total - true_e) < epsilon:
            return total
    raise RuntimeError(f"达到最大迭代次数 {max_iter}，仍未满足误差 {epsilon}")

def print_euler_sum_table(n: int) -> None:
    """
    打印 euler_infinite_sum(k) 的值，k 从 1 到 n，每行一个。

    参数:
        n: 正整数
    """
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

# 测试
print("euler_limit(1000) =", euler_limit(1000))
print("euler_infinite_sum(4) =", euler_infinite_sum(4))
print("euler_infinite_sum(1) =", euler_infinite_sum(1))
print("euler_infinite_sum(0) =", euler_infinite_sum(0))
print("euler_approx(1e-6) =", euler_approx(1e-6))
print("print_euler_sum_table(5):")
print_euler_sum_table(5)
```

**解释：**

- `euler_limit(n)` 直接套用 `(1 + 1/n)^n`。
- `euler_infinite_sum(n)` 累加 `1/k!`，k 从 0 到 n-1，对应“前 n 项”。
- `euler_infinite_sum(0)` 返回 `0`，`euler_infinite_sum(1)` 返回 `1`。
- `euler_approx(epsilon)` 用 while 不断累加，直到误差小于 `epsilon`。
- `print_euler_sum_table(n)` 逐行打印从 1 到 n 的近似值。

```python
import math

true_e = math.exp(1)
approx = euler_infinite_sum(10)

print(f"真实 e          = {true_e:.10f}")
print(f"前 10 项和近似  = {approx:.10f}")
print(f"绝对误差        = {abs(approx - true_e):.2e}")
```

**说明：**

- 真实 e 约为 `2.7182818285`。
- 用前 10 项和近似，误差已小于 `1e-7`。
- 项数越多，近似越精确。

## 5. 测试整数属性

1. 编写 `is_power_of_2(n)`，判断正整数 n 是否是 2 的幂。
   - 不使用 `math.sqrt`，只用除法和取模。
2. 编写 `is_power(b, n)`，判断 n 是否是 b 的幂。

```python
# 1. 判断 n 是否是 2 的幂
def is_power_of_2(n: int) -> bool:
    """
    判断正整数 n 是否为 2 的幂。

    参数:
        n: 正整数

    返回:
        True 表示 n 是 2 的幂，否则 False
    """
    while n % 2 == 0:
        n //= 2
    return n == 1

# 2. 判断 n 是否是 b 的幂
def is_power(b: int, n: int) -> bool:
    """
    判断正整数 n 是否为 b 的幂。

    参数:
        b: 底数，正整数
        n: 待判断的正整数

    返回:
        True 表示 n 是 b 的幂，否则 False
    """
    if n == 1:          # b^0 = 1，任何正整数的 0 次方都是 1
        return True
    if b == 1:          # 1 的任意次方都是 1，不能得到大于 1 的数
        return False
    while n % b == 0:
        n //= b
    return n == 1

# 测试 is_power_of_2
print("is_power_of_2(1):", is_power_of_2(1))
print("is_power_of_2(8):", is_power_of_2(8))
print("is_power_of_2(10):", is_power_of_2(10))

# 测试 is_power
print("is_power(2, 8):", is_power(2, 8))
print("is_power(3, 27):", is_power(3, 27))
print("is_power(4, 16):", is_power(4, 16))
print("is_power(4, 8):", is_power(4, 8))
```

**解释：**

- `is_power_of_2(n)`：不断除以 2，如果最后得到 1，说明 n 是 2 的幂。
- `is_power(b, n)`：不断除以 b，如果最后得到 1，说明 n 是 b 的幂。
- `n == 1` 时返回 `True`，因为任何正整数的 0 次方都是 1。
- `b == 1` 且 `n > 1` 时返回 `False`，因为 1 的任意次方都是 1，不可能得到大于 1 的数。
- 所有实现只用了除法和取模，符合题目要求。
